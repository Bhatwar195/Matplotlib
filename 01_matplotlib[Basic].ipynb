## Matplotlib Basic

#### Matplotlib is a Python library used for data visualization. It is commonly used to create line plots, bar charts, scatter plots, histograms, and other graphical representations.

```python
# importing lib 
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt 
```

### Define Sample Data

```python
x=[2,3,5,6,7,8,11,12]
y=[2010,2011,2012,2014,2016,2017,2019,2021]
```

### Basic Line Plot

```python
plt.plot(x,y)
plt.show()
```

### Line Plot with Marker and Style

```python
plt.plot(x, y,marker = 's',markeredgecolor = 'g',color = 'black',markersize = 6,linestyle='--')
#supported values are '-', '--', '-.', ':', 'None', ' ', '', 'solid', 'dashed', 'dashdot', 'dotted'       
plt.show()                           
```

### Line Plot with Labels and Title

```python
plt.plot(x, y, marker='o')
plt.xlabel("X Values")
plt.ylabel("Year")
plt.title("Line Plot Example")
plt.show()
```

### Save Plot as Image

```python
plt.plot(x, y, marker='o')
plt.xlabel("X Values")
plt.ylabel("Year")
plt.title("Saved Line Plot")
plt.savefig("line_plot.png")
plt.show()
```

```python
## save the graph as a file 
# the xlabel() and ylabel() functions to set a label for the x- and y-axis
# fontdict parameter in xlabel(), ylabel(), and title() to set font properties for the title and labels
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])
font1 = {'family':'serif','color':'r','size':15}
font2 = {'family':'serif','color':'y','size':15}
plt.plot(x, y,marker = '^',markeredgecolor = 'y',color = 'black',markersize = 10,linestyle='--')
plt.title("Sports Watch Data",fontdict=font1)
plt.xlabel("Average Pulse",fontdict=font2)
plt.ylabel("Calorie Burnage",fontdict=font2)
plt.legend(['Element_1'])  
plt.savefig('Figure_1_mat',dpi = 200)
plt.show()
```
