## Matplotlib Bar Chart

A bar chart is used to compare values across different categories.  
Each bar represents a category, and its height corresponds to the value.  

```python
import matplotlib.pyplot as plt
import numpy as np  
from matplotlib import style
```

### Define  sample Data

```python
classes = ["Python", "R", "AI", "ML", "DS"]
class1_students = [30, 10, 20, 25, 10] # out of 100 student in each class
class2_students = [40, 5, 20, 20, 10]
class3_students = [35, 5, 30, 15, 15]
```

### Simple Bar Chart

```python
plt.bar(classes, class1_students)
plt.show()
```

### Horizontal Bar Chart

```python
plt.barh(classes,class1_students)
plt.show()
```

### Styled Bar Chart

```python
plt.bar(classes, class1_students,width=0.8,color='orange',align='edge',edgecolor='r',alpha=0.9,label="Class 1 student",visible=True )
plt.show()
```

### ggplot ploting 
**ggplot** is a predefined plotting style in Matplotlib that improves the visual appearance of graphs by applying a clean, grid-based theme inspired by ggplot2 (R library).

```python
style.use("ggplot") 
plt.figure(figsize=(16,9)) # figure size with ratio 16:9
plt.bar(classes, class1_students, width = 0.6, align = "edge", color = "skyblue",edgecolor = "b", linewidth = 5, alpha = 0.9, linestyle = "--",label =" Class 1 Students")
plt.show()
```

```python
plt.figure(figsize=(13,9))
plt.barh(classes, class1_students, align = "edge", color = "skyblue",edgecolor = "b", linewidth = 5, alpha = 0.9, linestyle = "--",label =" Class 1 Students")
plt.show()
```

### Styled Bar Chart using ggplot

```python
style.use("ggplot") 
plt.figure(figsize=(16,9)) # figure size with ratio 16:9
plt.bar(classes, class1_students, width = 0.6, align = "edge", color = "pink",edgecolor = "m", linewidth = 5, alpha = 0.9, linestyle = "--",label =" Class 1 Students")
plt.title(" Bar Graph of V-tech classes")
plt.xlabel('Subjects',fontsize=15)
plt.ylabel('No.s of Students',fontsize=15)
plt.show()
```

### Grouped Bar Chart

```python
style.use("ggplot") 
plt.figure(figsize=(16,9)) # figure size with ratio 16:9
plt.bar(classes, class1_students, width = 0.6, align = "edge", color = "skyblue", linewidth = 5, alpha = 0.9, linestyle = "--",label =" Class 1 Students")
plt.bar(classes, class2_students, width = 0.6, align = "edge", color = "b", linewidth = 5, alpha = 0.9, linestyle = "--",label =" Class 1 Students")
plt.title(" Bar Graph of V-tech classes")
plt.xlabel('Subjects',fontsize=15)
plt.ylabel('No.s of Students',fontsize=15)
plt.show()
```
