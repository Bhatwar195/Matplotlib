## Matplotlib [Box-plot]

A box plot (box-and-whisker plot) is used to visualize the distribution of numerical data.  
It shows the median, quartiles, and potential outliers in the dataset.  

```python
import matplotlib.pyplot as plt
import numpy as np
```

### Generate Sample Data

```python
math_marks = [45, 50, 55, 60, 65, 70, 75, 71, 72,76,99]
science_marks = [40, 45, 50, 55, 60, 65, 70, 75, 80, 85]
english_marks = [35, 40, 45, 50, 55, 60, 65, 70, 75, 80]
```

### Basic Box Plot

```python
plt.boxplot(math_marks)
plt.show()
```

### Multiple Box Plot

```python
plt.boxplot([math_marks, science_marks, english_marks])
plt.show()
```

### Labeled Box Plot

```python
plt.boxplot(
    [math_marks, science_marks, english_marks],
    tick_labels=["Maths", "Science", "English"],
    patch_artist=True
)

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Marks Distribution using Box Plot")

plt.show()
```

### Styled Box Plot

```python
plt.boxplot(
    [math_marks, science_marks, english_marks],
    tick_labels=["Maths", "Science", "English"],
    patch_artist=True
)

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Styled Box Plot")

plt.show()
```
