### Matplotlib Scatter Plot

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.style 
```

```python
data=pd.read_csv("googleplaystore.csv",nrows=1000) ## Use only 1000 rows
data.head()
```

```python
data.shape
```

### simple Scatter Plot using Rating and Reviews

```python
plt.scatter(x=data['Rating'],y=data['Reviews'])
plt.show()
```

### Define x and y

```python
x=data['Rating']
y=data['Reviews']
```

### Styled Scatter plot

```python
plt.scatter(x,y,c='m',marker='*',s=100,alpha=0.5,linewidths=0.5,edgecolors='g')
plt.title("Google Play Store Apps Scatter plot")
plt.xlabel("Rating")
plt.ylabel("Reviews")
plt.show()
```

### Styled Scatter Plot

```python
plt.scatter(x,y,c='m',marker='*',s=100,alpha=0.5,linewidths=0.5,edgecolors='g')
plt.scatter(x,data['Installs'],c='b',marker='o',s=100,alpha=0.5,linewidths=0.5,edgecolors='y')
plt.title("Google Play Store Apps Scatter plot")
plt.xlabel("Rating")
plt.ylabel("Reviews & Install")
plt.show()
```
