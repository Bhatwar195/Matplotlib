## Matplotlib Pie Chart

 A **pie chart** is a circular statistical graphic that represents data in slices.  
Each slice shows the proportion or percentage of a category relative to the whole.  

```python
import matplotlib.pyplot as plt
import pandas as pd 
```

### Define Data

```python
classes = [
    "Python",
    "R",
    "Machine Learning",
    "Artificial Intelligence",
    "Data Science"
]

class1_students = [45, 15, 35, 25, 30]
```

### Pie Chart Configuration

```python
explode = [0.03, 0, 0.1, 0, 0]
colors = ["m", "b", "r", "y", "g"]
textprops = {"fontsize": 15}
```

### Simple Pie Chart

```python
plt.pie(class1_students, labels=classes)
plt.show()
```

### Styled Pie Chart

```python
plt.figure(figsize=(6,4))

plt.pie(
    class1_students,
    labels=classes,
    explode=explode,
    colors=colors,
    autopct="%0.2f%%",
    shadow=True,
    radius=1.4,
    startangle=270,
    textprops=textprops
)
plt.show()
```
