## Matplotlib [Histogram]

A **histogram** is used to represent the distribution of numerical data.  
It groups data into ranges called bins and displays their frequencies using bars.

```python
import matplotlib.pyplot as plt
import matplotlib.style 
import random
import pandas as pd
import numpy as np
```

### Generate Sample Data

```python
ml_students_age = np.random.randint(18,45, (100))
py_students_age = np.random.randint(15,40, (100))
print(ml_students_age)
print(py_students_age)
```

### Plotting Simple Histogram for Ml student age 

```python
plt.hist(ml_students_age,label = "ML_Student" ,color = "b",rwidth =0.9,histtype='bar')
font1 = {'family':'serif','color':'red','size':15}
font2 = {'family':'serif','color':'red','size':15}
plt.title("ML Students age histograms",fontdict = font1)
plt.xlabel("Students age cotegory",fontdict = font2)
plt.ylabel("No. Students age",fontdict = font2)
plt.legend(loc = 1)
plt.show()
```

### Histogram with Custom Bins

```python
bins = [18, 22, 26, 30, 34, 38, 42, 46]
plt.hist(ml_students_age, bins=bins)
plt.show()
```

### Plotting histogram of machine learning  and Python students age.

```python
bins = [15,20,25,30,35,40,45]
plt.hist([ml_students_age, py_students_age], bins, rwidth=0.8, histtype = "bar", orientation='vertical', 
         color = ["blue", "skyblue"], label = ["ML Student", "Py Student"])
plt.title("ML & PY Students age histograms",fontdict = font1)
plt.xlabel("Students age cotegory",fontdict = font2)
plt.ylabel("No. Students age",fontdict = font2)
plt.legend(loc=2)
plt.show()
```

### Styled Histogram

```python
plt.hist(
    ml_students_age,
    bins=bins,
    color='skyblue',
    edgecolor='black'
)
bins = [15,20,25,30,35,40,45]
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.title("Age Distribution of Students")
plt.show()
```
